```python
import cv2
import numpy as np
from sklearn.cluster import KMeans

# ---- Đường dẫn ảnh ----
input_path = "dog.jpg"
output_path = "dog_kmeans_25.jpg"

# ---- Đọc ảnh ----
img = cv2.imread(input_path)
h, w, c = img.shape
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# ---- Chuyển dữ liệu ảnh sang dạng 2D ----
pixels = img_rgb.reshape((-1, 3))

# ---- Xác định số cụm (k) để nén ảnh còn khoảng 25% dung lượng ----
# Mẹo: số cụm ~25% số màu ban đầu không hiệu quả, ta dùng k ≈ 64 hoặc 32 tùy kích thước ảnh
k = 64  # thử 64 màu, có thể giảm xuống 32 nếu muốn nhỏ hơn nữa

# ---- Dùng KMeans ----
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(pixels)

# ---- Lấy màu cụm và gán lại cho từng pixel ----
new_colors = kmeans.cluster_centers_.astype("uint8")
labels = kmeans.labels_
compressed_img = new_colors[labels].reshape((h, w, c))

# ---- Chuyển lại sang BGR để lưu ----
compressed_bgr = cv2.cvtColor(compressed_img, cv2.COLOR_RGB2BGR)
cv2.imwrite(output_path, compressed_bgr)

print(f" Ảnh đã được nén bằng KMeans và lưu tại: {output_path}")
```
